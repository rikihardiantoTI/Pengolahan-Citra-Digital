```python
img = cv2.imread("yuukimakoto (1).jpg")
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(img_rgb)
plt.title("Gambar Asli")
plt.axis("off")
plt.show()
```

```python
tinggi, lebar , channel = img_rgb.shape

print("Tinggi : ", tinggi)
print("Lebar : ", lebar)
print("Jumlah Channel : ", channel)
```

```python
R,G,B = cv2.split(img_rgb)

plt.figure(figsize=(12,4))
plt.subplot(1,3,1); plt.imshow(R, cmap="gray"); plt.title("Channel R"); plt.axis("off")
plt.subplot(1,3,2); plt.imshow(G, cmap="gray"); plt.title("Channel G"); plt.axis("off")
plt.subplot(1,3,3); plt.imshow(B, cmap="gray"); plt.title("Channel B"); plt.axis("off")
plt.show()
```

```python
img_gray = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)

plt.imshow(img_gray, cmap="gray")
plt.title("Gambar Grayscale")
plt.axis("off")
plt.show()
```

```python
def kuantisasi(gambar, tingkat):
    interval = 256 // tingkat
    hasil = (gambar // interval) * interval
    return hasil

img_q8 = kuantisasi(img_gray, 8)
img_q4 = kuantisasi(img_gray, 4)
img_q2 = kuantisasi(img_gray, 2)

plt.figure(figsize=(12,4))
plt.subplot(1,3,1); plt.imshow(img_q8, cmap="gray"); plt.title("Kuantisasi 8 Bit"); plt.axis("off")
plt.subplot(1,3,2); plt.imshow(img_q4, cmap="gray"); plt.title("Kuantisasi 4 Bit"); plt.axis("off")
plt.subplot(1,3,3); plt.imshow(img_q2, cmap="gray"); plt.title("Kuantisasi 2 Bit"); plt.axis("off")
plt.show()
```

```python
thresh_values = [50, 100, 150, 200]
hasil_biner = []

plt.figure(figsize=(10, 10))
for i, t in enumerate(thresh_values):
    ret, thresh = cv2.threshold(img_gray, t, 255, cv2.THRESH_BINARY)
    hasil_biner.append(thresh)
    plt.subplot(2, 2, i+1)
    plt.imshow(thresh, cmap='gray')
    plt.title(f'Biner - Threshold {t}')
    plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
# Menyimpan gambar grayscale dan gambar biner (hasil dari indeks ke-2 / Threshold 150)
cv2.imwrite("hasil_grayscale.jpg", img_gray)
cv2.imwrite("hasil_biner_thresh150.jpg", hasil_biner[2])

print("2 file berhasil disimpan di direktori Google Colab!")
```
